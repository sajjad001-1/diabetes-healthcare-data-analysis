```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

# Fetch dataset
cdc_diabetes_health_indicators = fetch_ucirepo(id=891)

# Features and target
X = cdc_diabetes_health_indicators.data.features
y = cdc_diabetes_health_indicators.data.targets

# Combine them into one DataFrame
df = pd.concat([X, y], axis=1)

print(df.head())
print(df.shape)
print(df.info())
```

```python
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
income_summary=(df.groupby('Income')['Diabetes_binary'].mean().reset_index(name='Diabetes_Percentage'))
income_summary['Diabetes_Percentage']*=100
sn.set_theme(style="whitegrid")
plt.figure(figsize=(10,6))
sn.barplot(x='Income',y='Diabetes_Percentage',data=income_summary,hue='Income',legend=False,palette="Blues_r")
plt.title('Diabetes Prevalence Decreases as Income Brackets Increases',fontsize=14,fontweight='bold')
plt.xlabel('Income Bracket(1=Lowest,8=Highest)',fontsize=12)
plt.ylabel('Percentage of People with Diabetes(%)',fontsize=12)
plt.tight_layout()
plt.savefig('Income_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
cardio_summary=df.groupby('Diabetes_binary')[['Stroke','HeartDiseaseorAttack']].mean()*100
cardio_summary=cardio_summary.reset_index()
cardio_melted=pd.melt(cardio_summary,id_vars=['Diabetes_binary'],value_vars=['Stroke','HeartDiseaseorAttack'],var_name='Condition',value_name='Percentage')
cardio_melted['Diabetes Status']=cardio_melted['Diabetes_binary'].map({0:'Non-Diabetic',1:'Diabetic'})
plt.figure(figsize=(10,6))
sn.barplot(x='Condition',y='Percentage',hue='Diabetes Status',data=cardio_melted,palette=['#4C72B0','#C44E52'])
plt.title('Cardiovascular Comorbidities by Diabetes Status',fontsize=14,fontweight='bold')
plt.xlabel('Medical Condition',fontsize=12)
plt.ylabel('Percentage of Population (%)',fontsize=12)
plt.legend(title='Patient Status')
plt.tight_layout()
plt.savefig('Cardiovascular_Commorbidities_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plot_df=df.copy()
plot_df['Diabetes']=plot_df['Diabetes_binary'].map({0:'Non-Diabetic',1:'Diabetic'})
sn.kdeplot(data=plot_df,x='BMI',hue='Diabetes',fill=True,common_norm=False,palette=['#4C72B0','#C44E52'],alpha=0.4,linewidth=2)
plt.axvline(x=30,color='black',linestyle='--',linewidth=1.5,label='Obesity Threshold (BMI 30)')
plt.title('BMI Distribution Comparison: Diabetics vs. Non-Diabetics',fontsize=14,fontweight='bold')
plt.xlabel('Body Mass Index(BMI)',fontsize=12)
plt.ylabel('Density',fontsize=12)
plt.xlim(10,60)
plt.legend(title='Patient Status')
plt.tight_layout()
plt.savefig('BMI_Distribution_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
walk_summary=df.groupby('Diabetes_binary')['DiffWalk'].mean().reset_index()
walk_summary['DiffWalk']*=100
walk_summary['Diabetes']=walk_summary['Diabetes_binary'].map({0:'Non-Diabetic',1:'Diabetic'})
plt.figure(figsize=(7,5))
sn.barplot(x='Diabetes',y='DiffWalk',data=walk_summary,hue='Diabetes',legend=False,palette=['#55A868','#C44E52'])
plt.title('Percentage of People Facing Difficulty Walking/Climbing Stairs',fontsize=14,fontweight='bold')
plt.xlabel('Patient Cohort',fontsize=12)
plt.ylabel('Struggle With Mobility (%)',fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plot_df=df.copy()
plot_df['Diabetes']=plot_df['Diabetes_binary'].map({0:'Non-Diabetic',1:'Diabetic'})
sn.boxplot(x='Diabetes',y='MentHlth',data=plot_df,hue='Diabetes',legend=False,palette=['#4C72B0','#8172B3'])
plt.title('Distribution of Poor Mental Health Days (Past 30 Days)',fontsize=14,fontweight='bold')
plt.xlabel('Patient Cohort',fontsize=12)
plt.ylabel('Number of Bad Mental Health Days (0-30)',fontsize=12)
plt.tight_layout()
plt.savefig('Distribution_MentalHlthDays_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
corr_matrix=df.corr()
plt.figure(figsize=(16,12))
sn.heatmap(corr_matrix,annot=True,cmap='coolwarm',vmin=-1,vmax=1,linewidths=0.5)
plt.title('CDC Diabetes Health Indicators: Full Feature Correlation Matrix',fontsize=16,fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plot_df=df.copy()
plot_df['Diabetes']=plot_df['Diabetes_binary'].map({0:'Non-Diabetic',1:'Diabetic'})
plot_df['Gender']=plot_df['Sex'].map({0:'Female',1:'Male'})
sn.histplot(data=plot_df,x='Gender',hue='Diabetes',multiple='dodge',shrink=0.7,discrete=True,palette=['#4C72B0','#C44E52'])
plt.title('Distribution of Diabetes Prevalence Across Genders',fontsize=14,fontweight='bold')
plt.xlabel('Gender',fontsize=12)
plt.ylabel('Number of Survey Respondents (Count)',fontsize=12)
plt.tight_layout()
plt.savefig('Distribution_Genders_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
diabetic_df=df[df['Diabetes_binary']==1]
health_labels={1.0:'1 - Excellent',2.0:'2 - Very Good',3.0:'3 - Good',4.0:'4 - Fair',5.0:'5 - Poor'}
health_counts=diabetic_df['GenHlth'].map(health_labels).value_counts().sort_index()
plt.figure(figsize=(8,8))
colors=sn.color_palette('pastel')[0:5]
explode=[0,0,0,0,0.1]
plt.pie(health_counts,labels=health_counts.index,autopct='%1.1f%%',startangle=140,colors=colors,explode=explode,shadow=True)
plt.title('Self-Reported General Health Rating Among Diabetic Individuals',fontsize=14,fontweight='bold')
plt.tight_layout()
plt.savefig('Pie_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```
